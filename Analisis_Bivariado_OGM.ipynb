# Iniciemos con el Análisis Bivariado

En este punto tenemos identificado el conjunto de datos y sus componentes.
Ahora Vamos a utilizar las librerías de Python: Matplotlib y seaborn para graficar los registros del conjunto de datos cargado "Domestic Violence in Colombia" y realizar asi el analsis bivariado sobre este conjunto de datos.

```python
# Incluyamos la Librería Altair para visualización estadistica del
# conjunto de datos que etamos analizando: "Violencia Domestica en Colombia del año 2010 al 2021"
import altair as alt
```

```python
# Vamos a genera un grafico de dispersion que incluya: Departamentos y Nro. de casos reportados por Departamento.
# Intanciar un objeto Chart con el dataframe df como parámetro
base = alt.Chart(df)

# Llamar el método mark_circle() para especificar que va a trazar un diagrama de dispersión (scatter plot)
chart = base.mark_circle()

# Especificar el nombre de las columnas:
chart.encode(x='DEPARTAMENTO', y='CANTIDAD')
```

# Análisis de la Gráfica
De la gráfica anterior podemos decir:
Hay tres departamentos que reportan el mayor numero de casos de violencia:
- Cundinamarca
- Valle
- Meta

```python
# Generemos un Histograma por cantidad de reportes de casos de violencia por
# Departamento
alt.Chart(df).mark_bar().encode(alt.X('DEPARTAMENTO'), y='count()')
```

# Análisis de la Gráfica
De la gráfica de barras anterior, donde se muestra por Departamento el número de reportes de casos de Violencia, podemos mencionar los 5 primeros departamentos con mas numero de reportes:

1. Valle
2. Santander
3. Boyacá
4. Cundinamarca
5. Antioquia

```python
# Generemos para el Departamento del Valle del Cauca
# las armas mas usadas en los casos de violencia reportados

#Importamos la libreria matplotlib & seaborn
import matplotlib.pyplot as plt
import seaborn as sns

# Generemos del conjunto de datos el listado de armas mas usadas en los casos
# de violencia en orden ascendente
cantidad_armas = df['ARMAS MEDIOS'].value_counts().sort_values(ascending=False)

# Grafiquemos del listado de armas mas usadas
# en los casos de violencia reportados del Departamento del Valle del Cauca
ax = sns.countplot(x='ARMAS MEDIOS', data=df[df['DEPARTAMENTO'] == 'VALLE'], order=cantidad_armas.index)
ax.set_xticklabels(ax.get_xticklabels(), rotation=90)
plt.show()
```

# Análisis de la Gráfica
De la gráfica de barras anterior, se muestra que las armas contundentes son las que predomina en los casos de Violencia para el Departamento del Valle, seguido de las armas blancas o cortopunzantes.

```python
# Generemos un Histograma por cantidad de tipos de arma o medio usado
# en los casos de violencia domestica reportada.
alt.Chart(df).mark_bar().encode(alt.X('ARMAS MEDIOS'), y='count()')
```

# Análisis de la Gráfica
De la gráfica de barras anterior, donde se muestra el tipo de arma o medio usado en los casos reportados de violencia, se puede establecer el Top 3 de las mas usadas:

1. Contundentes
2. No Reportada
3. Arma Blanca / Cortopunzante


```python
# Generemos un Histograma por genero
# en los casos de violencia domestica reportada.
alt.Chart(df).mark_bar().encode(alt.X('GENERO'), y='count()')
```

# Análisis de la Gráfica
De la gráfica de barras anterior, donde se muestra el género reportado en los casos de Violencia, es notorio que es a las Mujeres (Género Femenino) a quienes se les ataca mayormente en los casos de violencia reportados.



```python
# Obtener la correlación entre variables por medio del coeficiente de correlación de Pearson
correlation = df.corr(method='pearson',numeric_only=True).reset_index().melt('index')
correlation.columns = ['variable1', 'variable2', 'indice-correlation']

correlation.head()
```

```python
chart = alt.Chart(correlation).mark_rect().encode(
    x=alt.X('variable1', title=None),
    y=alt.Y('variable2', title=None),
    # El color de cada casilla dependerá del índice de correlación hallado
    color=alt.Color('indice-correlation'),
).properties(
    # Establecemos el tamaño del gráfico
    width=alt.Step(40),
    height=alt.Step(40)
)

# Personalizamos el tamaño del texto de las casillas internas
chart += chart.mark_text(size=10).encode(
    # Establecemos el formato de la barra de temperatura
    text=alt.Text('indice-correlation', format=".2f"),
    color=alt.condition(
        "datum.correlation > 0.5",
        alt.value('white'),
        alt.value('black')
    )
)

# Desplegamos el gráfico
chart
```

# Análisis de la Gráfica
El Mapa de Calor, muestra que hay una correlación muy debil o diria que Nula entre el Codigo DANE, que representa la identificiación del Departamento como código en el Departamento Administrativo Nacional de Estadística y el numero de casos de Violencia. Lo cual tiene mucho sentido, pues si bien no hay una relación directa, si otros factores como: Nivel Educativo, estrato, indice de pobreza, entre otros, pueden ser los verdaderas variable correlacionadas con los casos de violencia mostrados en los Municipios. Por lo que era de esperarse que no se evidenciara una correlacion entre estas dos (2) variables.
